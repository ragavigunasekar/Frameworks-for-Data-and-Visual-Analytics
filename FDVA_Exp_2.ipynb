```python
import pandas as pd
import sqlite3

# --------------------------------------------------
# 1. IMPORT DATA FROM CSV
# --------------------------------------------------

df_csv = pd.read_csv("/content/test_Y3wMUE5_7gLdaTN.csv")

print("CSV DATA:")
display(df_csv.head())


# --------------------------------------------------
# 2. IMPORT DATA FROM EXCEL
# --------------------------------------------------

df_excel = pd.read_excel(
    "/content/Historicalinvesttemp.xlsx",
    sheet_name="Sheet1"
)

print("EXCEL DATA:")
display(df_excel.head())


# --------------------------------------------------
# 3. IMPORT DATA FROM SQL DATABASE
# --------------------------------------------------

# Create a temporary SQLite database
connection = sqlite3.connect("sample.db")

# Store Excel data into SQL table
df_excel.to_sql(
    "data_table",
    connection,
    if_exists="replace",
    index=False
)

# Read data from SQL database
df_sql = pd.read_sql(
    "SELECT * FROM data_table",
    connection
)

print("SQL DATA:")
display(df_sql.head())

connection.close()


# --------------------------------------------------
# 4. IMPORT DATA FROM WEB
# --------------------------------------------------

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df_web = pd.read_csv(url)

print("WEB DATA:")
display(df_web.head())


# --------------------------------------------------
# 5. EXPORT DATAFRAME TO EXCEL
# --------------------------------------------------

output_file = "/content/exported_data.xlsx"

df_csv.to_excel(
    output_file,
    index=False
)

print("DataFrame exported successfully!")
print("File:", output_file)
```
